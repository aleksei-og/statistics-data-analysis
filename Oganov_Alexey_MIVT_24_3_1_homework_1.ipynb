1. Объясните разницу между выборочным и генеральным средним. Когда применяется доверительный интервал?

Разница между выборочным и генеральным средним заключается в том, что они относятся к разным группам данных и применяются в разных контекстах:

**Генеральное среднее** - это среднее значение, рассчитанное для всей популяции.
Генеральное среднее — это истинное среднее для всей совокупности данных.
Например, если мы измеряем рост всех студентов в университете, то генеральное среднее — это средний рост всех студентов университета.

**Выборочное среднее** - это среднее значение, рассчитанное на основе выборки данных из популяции.
Выборочное среднее — это приближенная оценка генерального среднего, так как оно рассчитывается только на основе части данных, а не на основе всей популяции.
Например, если мы измеряем рост только у 50 студентов из 1000, то это выборка, и выборочное среднее будет представлять собой средний рост только этих 50 студентов, что является приблизительной оценкой среднего роста всей популяции студентов.

**Доверительный интервал** применяется, когда мы хотим оценить диапазон значений, в котором с определенной вероятностью (например, 95% или 99%) находится истинное значение параметра (например, среднее значение) генеральной популяции. Он используется в следующих случаях:

При оценке среднего значения на основе выборки.
При оценке других статистических параметров, таких как пропорции, дисперсии и т.д.
Когда важна оценка точности полученных данных или результатов.

Пример: Если выборочное среднее для какой-то величины составляет 50, а доверительный интервал от 45 до 55, это значит, что с вероятностью 95% генеральное среднее лежит в этом интервале.

#### Задача 1 (10 баллов)
Дан интервальный ряд (загружен в файл `interval_data.csv`):

1. Постройте гистограмму распределения.
2. Рассчитайте среднее значение, моду и медиану с помощью встроенных методов и написанных функций собственноручно.

```python
### Импорты необходимых библиотек
import pandas as pd
import numpy as np
```

```python
# загрузка данных из CSV файла
df = pd.read_csv('interval_data.csv')

# разделяю интервалы на нижнюю и верхнюю границу
df['Lower'] = df['Interval'].apply(lambda x: int(x.split('-')[0]))
df['Upper'] = df['Interval'].apply(lambda x: int(x.split('-')[1]))

# рассчитываю среднюю точку интервала
df['Midpoint'] = (df['Lower'] + df['Upper']) / 2

# 1. среднее значение (встроенный метод)
mean_builtin = np.average(df['Midpoint'], weights=df['Frequency'])
print(f"Среднее значение (встроенный метод): {mean_builtin}")

# 2. мода (интервал с максимальной частотой)
mode_builtin_interval = df.loc[df['Frequency'].idxmax()]['Interval']
print(f"Мода (встроенный метод): Интервал {mode_builtin_interval}")

# 3. медиана (встроенный метод)
cumulative_freq = df['Frequency'].cumsum()
median_builtin_interval = df[cumulative_freq >= cumulative_freq.iloc[-1] / 2].iloc[0]['Interval']
print(f"Медиана (встроенный метод): Интервал {median_builtin_interval}")


# собственноручно написанные функции

# функция для расчета среднего значения
def mean_custom(midpoints, frequencies):
    return np.sum(midpoints * frequencies) / np.sum(frequencies)

# функция для нахождения моды (интервал с максимальной частотой)
def mode_custom(frequencies, intervals):
    mode_index = np.argmax(frequencies)  # Находим индекс максимальной частоты
    return intervals[mode_index]

# функция для вычисления медианы
def median_custom(frequencies, intervals, midpoints):
    cumulative_freq = np.cumsum(frequencies)
    total_freq = np.sum(frequencies)
    median_index = np.where(cumulative_freq >= total_freq / 2)[0][0]

    # медиана находится в интервале с кумулятивной частотой, превышающей половину от общей частоты
    lower_bound = intervals[median_index][0]  # нижняя граница интервала
    freq_before = cumulative_freq[median_index] - frequencies[median_index]  # кумулятивная частота до интервала
    interval_length = intervals[median_index][1] - intervals[median_index][0]  # длина интервала
    median = lower_bound + ((total_freq / 2 - freq_before) / frequencies[median_index]) * interval_length
    return median

# применение собственноручных функций

# среднее значение
mean_custom_result = mean_custom(df['Midpoint'], df['Frequency'])
print(f"Среднее значение (собственная функция): {mean_custom_result}")

# мода
mode_custom_result = mode_custom(df['Frequency'], df['Interval'])
print(f"Мода (собственная функция): Интервал {mode_custom_result}")

# медиана
intervals_tuple = [(int(i.split('-')[0]), int(i.split('-')[1])) for i in df['Interval']]
median_custom_result = median_custom(df['Frequency'], intervals_tuple, df['Midpoint'])
print(f"Медиана (собственная функция): {median_custom_result}")
```

#### Задача 2 (15 баллов)
Сгенерируйте выборку объёма 1000 из нормального распределения с параметрами $\mu = 5$, $\sigma = 2$.

1. Постройте графики PDF и CDF.
2. Найдите 95% доверительный интервал для среднего значения.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# параметры нормального распределения
mu = 5
sigma = 2
sample_size = 1000

# генерация выборки из нормального распределения
sample = np.random.normal(mu, sigma, sample_size)

# 1. строю графики PDF и CDF

# для PDF
x_values = np.linspace(min(sample), max(sample), 1000)
pdf_values = stats.norm.pdf(x_values, mu, sigma)

# для CDF
cdf_values = stats.norm.cdf(x_values, mu, sigma)

# график PDF
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(x_values, pdf_values, label='Теоретическая PDF', color='blue')
plt.hist(sample, bins=30, density=True, alpha=0.6, color='green', label='Выборка')
plt.title("График PDF")
plt.xlabel("x")
plt.ylabel("Плотность вероятности")
plt.legend()

# график CDF
plt.subplot(1, 2, 2)
plt.plot(x_values, cdf_values, label='Теоретическая CDF', color='blue')
plt.hist(sample, bins=30, cumulative=True, density=True, alpha=0.6, color='green', label='Выборка')
plt.title("График CDF")
plt.xlabel("x")
plt.ylabel("Кумулятивная вероятность")
plt.legend()

plt.tight_layout()
plt.show()

# 2. ищу 95% доверительный интервал для среднего значения выборки

# оценка среднего и стандартной ошибки
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)
std_error = sample_std / np.sqrt(sample_size)

# 95% доверительный интервал для среднего значения
confidence_level = 0.95
z_value = stats.norm.ppf(1 - (1 - confidence_level) / 2)

# погрешность и доверительный интервал
margin_of_error = z_value * std_error
ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

print(f"95% доверительный интервал для среднего значения: ({ci_lower:.2f}, {ci_upper:.2f})")
```

#### Задача 4 (15 баллов)
Проверьте гипотезу $H_0: \mu = 25$ против $H_a: \mu \neq 25$ для выборки размера 50 с параметрами $\mu = 23$, $\sigma = 5$ при уровне значимости $\alpha = 0.05$.

```python
import numpy as np
import scipy.stats as stats

# параметры
mu_population = 23  # истинное среднее в генеральной совокупности
sigma_population = 5  # стандартное отклонение в генеральной совокупности
sample_size = 50  # размер выборки
alpha = 0.05  # уровень значимости

# генерация выборки из нормального распределения
sample = np.random.normal(mu_population, sigma_population, sample_size)

# рассчитываю выборочное среднее и стандартную ошибку
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)  # стандартное отклонение выборки
std_error = sample_std / np.sqrt(sample_size)  # стандартная ошибка

# рассчитываю t-статистику
t_statistic = (sample_mean - 25) / std_error  # H0: mu = 25

# определяю критическое значение для двустороннего теста
t_critical = stats.t.ppf(1 - alpha / 2, df=sample_size - 1)

# рассчитываю p-значение
p_value = 2 * (1 - stats.t.cdf(abs(t_statistic), df=sample_size - 1))  # двусторонний тест

# вывожу результаты
print(f"Среднее выборки: {sample_mean:.2f}")
print(f"t-статистика: {t_statistic:.2f}")
print(f"Критическое значение t: {t_critical:.2f}")
print(f"p-значение: {p_value:.4f}")

# проверка гипотезы
if p_value < alpha:
    print("Отвергаем нулевую гипотезу H0: Среднее не равно 25.")
else:
    print("Не отвергаем нулевую гипотезу H0: Среднее равно 25.")
```

### Часть 3: Анализ данных (25 баллов)

Используя датасет `angled_blob_data.csv`:

1. Постройте график распределения точек.
2. Примените линейную регрессию и определите уравнение линии.
3. Вычислите $R^2$. Интерпретируйте полученный результат.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# загрузка данных
df = pd.read_csv('angled_blob_data.csv')

# 1. построение графика распределения точек
plt.figure(figsize=(8, 6))
plt.scatter(df['x'], df['y'], color='blue', alpha=0.6)
plt.title("График распределения точек")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()

# 2. линейная регрессия
X = df[['x']]  # признак (независимая переменная)
y = df['y']    # целевая переменная (зависимая переменная)

# создание модели линейной регрессии
model = LinearRegression()

# обучение модели
model.fit(X, y)

# получаем коэффициенты регрессии
slope = model.coef_[0]  # коэффициент наклона
intercept = model.intercept_  # свободный член

# уравнение прямой
print(f"Уравнение прямой: y = {slope:.2f}x + {intercept:.2f}")

# 3. вычисление R^2
y_pred = model.predict(X)  # предсказанные значения
r2 = r2_score(y, y_pred)  # коэффициент детерминации

print(f"R^2: {r2:.4f}")

# интерпретация
if r2 > 0.8:
    print("Модель хорошо объясняет вариацию данных (высокая степень корреляции).")
elif r2 > 0.5:
    print("Модель объясняет умеренную часть вариации данных.")
else:
    print("Модель объясняет небольшую часть вариации данных.")
```
