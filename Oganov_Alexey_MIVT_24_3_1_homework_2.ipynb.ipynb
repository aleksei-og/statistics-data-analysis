**№1.**

**Нулевая гипотеза** (H₀) — это исходное утверждение о генеральной совокупности, которое предполагает отсутствие эффекта или различий, или что определённый параметр равен некоторому значению. Например, если проверяется влияние лекарства, то нулевая гипотеза может быть такой: "лекарство не оказывает эффекта на здоровье".

**Альтернативная гипотеза** (H₁) — это гипотеза, противоположная нулевой, которая предполагает наличие эффекта или различий, или что параметр отличается от определённого значения. В примере с лекарством альтернативная гипотеза может утверждать, что "лекарство оказывает эффект на здоровье".

Ошибки при проверке гипотез:

**Ошибка первого рода** (α) — это ошибка, при которой отвергается нулевая гипотеза, хотя на самом деле она верна. То есть, исследователь приходит к выводу, что есть эффект (например, что лекарство влияет на здоровье), хотя на самом деле его нет.

**Ошибка второго рода** (β) — это ошибка, при которой не отвергается нулевая гипотеза, хотя она на самом деле ложна. То есть, исследователь не обнаруживает эффекта (например, что лекарство не оказывает влияния), хотя оно существует.

Эти ошибки имеют разные последствия:

Ошибка первого рода — может привести к ложным заключениям о наличии эффекта, что может повлиять на принятие решений, например, в медицине.
Ошибка второго рода — может привести к невыявлению истинного эффекта, что также может иметь негативные последствия, например, если эффективное лечение будет отвергнуто.

**№2**

**Уровень значимости** (α) — это заранее выбранная вероятность того, что при истинности нулевой гипотезы мы её отвергнем. Он устанавливает порог для принятия решения об отклонении нулевой гипотезы, и его обычно выбирают до начала исследования. Например, уровень значимости 0,05 означает, что исследователь готов принять вероятность ошибки первого рода в 5%, то есть вероятность того, что нулевая гипотеза будет отвергнута, хотя она на самом деле верна.
Связь уровня значимости с ошибками первого и второго рода:

**Ошибка первого рода** (α):
Уровень значимости напрямую связан с вероятностью ошибки первого рода. Ошибка первого рода происходит, когда мы отвергаем нулевую гипотезу, хотя она на самом деле верна.
Например, при уровне значимости 0,05 вероятность ошибки первого рода составляет 5%. Это означает, что из 100 случаев, когда нулевая гипотеза верна, мы будем отвергать её 5 раз (ошибочно).

**Ошибка второго рода** (β):
Ошибка второго рода происходит, когда мы не отвергаем нулевую гипотезу, хотя она на самом деле ложна.
Уровень значимости (α) и ошибка второго рода (β) находятся в обратной зависимости: чем меньше мы выбираем α (строже устанавливаем порог для отклонения гипотезы), тем больше вероятность ошибки второго рода. Это связано с тем, что уменьшение α (строже проверка на значимость) повышает вероятность того, что мы не обнаружим истинного эффекта, если он существует.

**Пример:**

Предположим, что мы тестируем новый метод лечения с уровнем значимости 0,05:

Ошибка первого рода: Если уровень значимости α = 0,05, то вероятность того, что мы отвергнем нулевую гипотезу (о том, что метод не эффективен), хотя на самом деле она верна, составит 5%.
Ошибка второго рода: Если уровень значимости слишком низкий (например, α = 0,01), вероятность ошибки второго рода (не обнаружить эффективность метода, если она существует) увеличится.

**№3**

**Критическая область** — это часть распределения статистики, в которой нулевая гипотеза отвергается. Определяется эта область с помощью выбранного уровня значимости (α), который устанавливает порог для принятия решения о том, что наблюдаемое значение статистики достаточно экстремально, чтобы отклонить нулевую гипотезу.
**Построение критической области:**

* Выбор уровня значимости (α): Обычно выбирается значение α (например, 0,05 или 0,

01), которое определяет вероятность ошибки первого рода (ошибка отвергания истинной нулевой гипотезы).

* Определение типа теста: В зависимости от гипотезы тест может быть:
Двусторонним: когда рассматриваются как большие, так и малые отклонения от предполагаемого значения.
Односторонним: когда рассматривается отклонение в одном направлении (либо больше, либо меньше).

* Выбор статистики: В случае с тестом Стьюдента для выборки средней мы используем t-статистику.

* Определение критической области:
Для двустороннего теста критическая область будет располагаться в обоих хвостах распределения. То есть, мы отвергаем нулевую гипотезу, если наблюдаемое значение t-статистики попадает в верхний или нижний хвост распределения, каждый из которых ограничен значением, соответствующим уровню значимости.
Для одностороннего теста критическая область будет расположена только в одном хвосте (верхнем или нижнем).

* Сравнение наблюдаемого значения t с критической областью: Если t-статистика попадает в критическую область, мы отвергаем нулевую гипотезу. Если она не попадает в эту область, мы не отвергаем нулевую гипотезу.

**№4**

**Критерий согласия Пирсона** — это статистический тест, который используется для оценки того, насколько хорошо наблюдаемые данные соответствуют ожидаемым значениям, исходя из гипотезы о распределении данных. Критерий Пирсона проверяет гипотезу о том, что данные не противоречат предполагаемому распределению или не различаются по группам.

**Основные случаи применения** критерия Пирсона:

* Проверка соответствия наблюдаемых данных теоретическому распределению:
Этот тест применяется, когда мы хотим узнать, насколько хорошо данные следуют заданному распределению, например, нормальному распределению, биномиальному или равномерному.
Пример: Проверка того, насколько результат бросков игральной кости соответствует равномерному распределению.

* Проверка независимости между категориальными переменными:
Этот тест применяется для проверки гипотезы о независимости двух категориальных переменных в таблице сопряженности (например, в таблице "частоты" или "сопоставления").
Пример: Проверка гипотезы о том, что пол человека не влияет на выбор его профессии (то есть, пол и профессия независимы).

χ2=(8−10)/10+(12−10)/10+(11−10)/10+(9−10)/10+(10−10)/10+(10−10)/10

Вычислим:
χ2​=1.

#### Задача 1

Проверьте гипотезу о том, что среднее значение выборки равно 50, используя следующие данные:

```
[48, 51, 50, 49, 50, 52, 48, 47, 53, 50]
```
1. Рассчитайте \( t \)-статистику.
2. Проверьте гипотезу на уровне значимости \($\alpha$ = 0.05\).

````python
import numpy as np
from scipy import stats

data = np.array([48, 51, 50, 49, 50, 52, 48, 47, 53, 50])

# 1. рассчитываю t-статистику
sample_mean = np.mean(data)  # среднее значение выборки
mu_0 = 50  # гипотетическое среднее значение
sample_std = np.std(data, ddof=1)  # стандартное отклонение выборки
n = len(data)  # размер выборки

# t-статистика
t_stat = (sample_mean - mu_0) / (sample_std / np.sqrt(n))

# 2. проверяю гипотезу на уровне значимости alpha = 0.05
alpha = 0.05

# степени свободы
df = n - 1

# критическое значение t для двустороннего теста
t_critical = stats.t.ppf(1 - alpha / 2, df)

# p-значение для двустороннего теста
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df))

# вывод результатов
print("t-статистика:", t_stat)
print("Критическое значение t:", t_critical)
print("p-значение:", p_value)

# проверка гипотезы
if abs(t_stat) > t_critical:
    print("Отвергаем нулевую гипотезу")
else:
    print("Не отвергаем нулевую гипотезу")
````

#### Задача 2

Используя следующий массив:

```
[76, 6, 34, 42, 36, 50, 51, 39, 18, 99, 93, 48, 74, 39, 64, 87, 17, 23, 35, 15]
```

1. Проверьте, является ли эта выборка случайной, используя метод последовательностей.
2. Примите или отклоните гипотезу о случайности выборки.

````python
import numpy as np

data = [76, 6, 34, 42, 36, 50, 51, 39, 18, 99, 93, 48, 74, 39, 64, 87, 17, 23, 35, 15]

# 1. рассчитываю среднее значение выборки
mean_value = np.mean(data)

# 2. применяю критерий: больше или меньше среднего
# если число больше среднего - это "успех", если меньше - "неудача"
binary_data = [1 if x > mean_value else 0 for x in data]

# 3. подсчитываю количество последовательностей
sequences = 1  # первая последовательность начинается с первого элемента
for i in range(1, len(binary_data)):
    if binary_data[i] != binary_data[i - 1]:
        sequences += 1

# ожидаемое количество последовательностей для случайных данных
expected_sequences = len(data) / 2  # для случайных данных вероятность успеха 0.5

# 4. вывожу результаты
print("Среднее значение выборки:", mean_value)
print("Бинарные данные (0 - меньше среднего, 1 - больше среднего):", binary_data)
print("Количество последовательностей:", sequences)
print("Ожидаемое количество последовательностей для случайных данных:", expected_sequences)

# проверка гипотезы:
alpha = 0.05
# поскольку у нас нет точной формулы для проверки по методу последовательностей, можно
# считать, что если количество последовательностей близко к ожидаемому, то гипотеза о случайности
# не отклоняется. Если же количество сильно отличается от ожидаемого, то гипотеза о случайности
# отвергается.

if abs(sequences - expected_sequences) > 2:
    print("Гипотеза о случайности выборки отклоняется.")
else:
    print("Гипотеза о случайности выборки не отклоняется.")
````

#### Задача 3

Проверьте гипотезу о честности игрального кубика. Используйте сгенерированные данные бросков кубика (N=100).
1. Постройте гистограмму частот выпадения каждой грани.
2. Примените критерий согласия Пирсона для проверки гипотезы \(H_0\): кубик честный (\(p = 1/6\)).

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# генерирую данные бросков кубика (N=100)
np.random.seed(42)
rolls = np.random.randint(1, 7, 100)  # генерирую 100 бросков (от 1 до 6)

# 1. строю гистограмму частот выпадений каждой грани
plt.hist(rolls, bins=np.arange(0.5, 7.5, 1), edgecolor='black', rwidth=0.8)
plt.xlabel('Грань кубика')
plt.ylabel('Частота выпадений')
plt.title('Гистограмма частот выпадений каждой грани кубика')
plt.xticks(np.arange(1, 7))
plt.show()

# 2. применяю критерий согласия Пирсона
# ожидаемая частота для каждой грани (если кубик честный)
expected_frequency = 100 / 6

# подсчитываю наблюдаемую частоту выпадений каждой грани
observed_frequency = [np.sum(rolls == i) for i in range(1, 7)]

# применяю критерий согласия Пирсона
chi_squared_statistic, p_value = stats.chisquare(observed_frequency, f_exp=[expected_frequency] * 6)

# вывод результатов
print(f"Наблюдаемая частота: {observed_frequency}")
print(f"Ожидаемая частота: {expected_frequency:.2f}")
print(f"Статистика хи-квадрат: {chi_squared_statistic:.2f}")
print(f"p-значение: {p_value:.4f}")

# проверка гипотезы
alpha = 0.05
if p_value < alpha:
    print("Отвергаем нулевую гипотезу: кубик не честный.")
else:
    print("Не отвергаем нулевую гипотезу: кубик честный.")
````

1. Рассчитайте \(t\)-статистику для двух выборок.
2. Проверьте гипотезу на уровне значимости \(lpha = 0.01\).

````python
import numpy as np
from scipy import stats

sample1 = np.array([10.1, 9.8, 10.5, 10.2, 9.9, 10.0, 10.3, 9.7])
sample2 = np.array([11.0, 10.8, 11.1, 10.7, 11.2, 10.9, 11.1, 10.6])

# 1. рассчитываю средние значения выборок
mean1 = np.mean(sample1)
mean2 = np.mean(sample2)

# 2. рассчитываю стандартные отклонения выборок
std1 = np.std(sample1, ddof=1)
std2 = np.std(sample2, ddof=1)

# 3. рассчитываю размеры выборок
n1 = len(sample1)
n2 = len(sample2)

# 4. рассчитываю t-статистику
t_stat = (mean1 - mean2) / np.sqrt((std1**2 / n1) + (std2**2 / n2))

# 5. рассчитываю степени свободы для t-теста (формула для независимых выборок)
df = ((std1**2 / n1 + std2**2 / n2)**2) / (((std1**2 / n1)**2 / (n1 - 1)) + ((std2**2 / n2)**2 / (n2 - 1)))

# 6. рассчитываю p-значение для двустороннего теста
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df))

# 7. проверяю гипотезу на уровне значимости alpha = 0.01
alpha = 0.01

# вывод результатов
print(f"Среднее значение выборки 1: {mean1:.2f}")
print(f"Среднее значение выборки 2: {mean2:.2f}")
print(f"t-статистика: {t_stat:.2f}")
print(f"Степени свободы: {df:.2f}")
print(f"p-значение: {p_value:.4f}")

# проверка гипотезы
if p_value < alpha:
    print("Отвергаем нулевую гипотезу: средние выборок различаются.")
else:
    print("Не отвергаем нулевую гипотезу: средние выборок не различаются.")
````

#### Задача 5

Для 200 бросков нечестного кубика известно, что распределение вероятностей выпадения каждой грани:
```
[0.1, 0.15, 0.2, 0.25, 0.2, 0.1]
```

1. Сгенерируйте данные для 200 бросков.
2. Проверьте гипотезу о соответствии распределения с указанными вероятностями, используя критерий согласия Пирсона.

````python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# 1. генерирую данные для 200 бросков с заданными вероятностями
np.random.seed(42)
probabilities = [0.1, 0.15, 0.2, 0.25, 0.2, 0.1]
rolls = np.random.choice([1, 2, 3, 4, 5, 6], size=200, p=probabilities)

# 2. подсчитываю наблюдаемую частоту для каждой грани
observed_frequency = [np.sum(rolls == i) for i in range(1, 7)]

# 3. ожидаемая частота для каждой грани (200 бросков)
expected_frequency = [200 * p for p in probabilities]

# 4. применяю критерий согласия Пирсона
chi_squared_statistic, p_value = stats.chisquare(observed_frequency, f_exp=expected_frequency)

# 5. строю гистограмму для визуализации частот
plt.hist(rolls, bins=np.arange(0.5, 7.5, 1), edgecolor='black', rwidth=0.8)
plt.xlabel('Грань кубика')
plt.ylabel('Частота выпадений')
plt.title('Гистограмма частот выпадений каждой грани (200 бросков)')
plt.xticks(np.arange(1, 7))
plt.show()

# 6. вывод результатов
print(f"Наблюдаемая частота: {observed_frequency}")
print(f"Ожидаемая частота: {expected_frequency}")
print(f"Статистика хи-квадрат: {chi_squared_statistic:.2f}")
print(f"p-значение: {p_value:.4f}")

# проверка гипотезы
alpha = 0.05
if p_value < alpha:
    print("Отвергаем нулевую гипотезу: распределение не соответствует заданным вероятностям.")
else:
    print("Не отвергаем нулевую гипотезу: распределение соответствует заданным вероятностям.")
````
